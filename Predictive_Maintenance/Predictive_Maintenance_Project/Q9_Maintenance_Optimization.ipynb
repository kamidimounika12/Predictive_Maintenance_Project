```python
#Load Data
from pyspark.sql.functions import *

maint = spark.table(
    "workspace.predictive_maintenance.silver_maint"
)

failures = spark.table(
    "workspace.predictive_maintenance.silver_failures"
)
```

```python
#Monthly Maintenance Count
maintenance_monthly = maint.withColumn(
    "year_month",
    date_format("datetime","yyyy-MM")
).groupBy(
    "year_month"
).count().orderBy(
    "year_month"
)

display(maintenance_monthly)
```

```python
#Monthly Failure Count
failure_monthly = failures.withColumn(
    "year_month",
    date_format("datetime","yyyy-MM")
).groupBy(
    "year_month"
).count().orderBy(
    "year_month"
)

display(failure_monthly)
```

```python
#Maintenance by Component
maintenance_component = maint.groupBy(
    "comp"
).count().orderBy(
    col("count").desc()
)

display(maintenance_component)
```

```python
#Failure by Component
failure_component = failures.groupBy(
    "failure"
).count().orderBy(
    col("count").desc()
)

display(failure_component)
```

```python
#Maintenance Summary
print("Total Maintenance Records:", maint.count())
print("Total Failure Records:", failures.count())
```
