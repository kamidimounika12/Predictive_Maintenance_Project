```python
#Load Gold Table
from pyspark.sql.functions import avg

df = spark.table(
    "workspace.predictive_maintenance.gold_machine_telemetry"
)
```

```python
#Average Telemetry by Age Group
age_analysis = df.groupBy(
    "age_group"
).agg(
    avg("volt").alias("avg_volt"),
    avg("rotate").alias("avg_rotate"),
    avg("pressure").alias("avg_pressure"),
    avg("vibration").alias("avg_vibration")
)

display(age_analysis)
```

```python
#Analysis by Machine Model
model_analysis = df.groupBy(
    "model"
).agg(
    avg("volt").alias("avg_volt"),
    avg("rotate").alias("avg_rotate"),
    avg("pressure").alias("avg_pressure"),
    avg("vibration").alias("avg_vibration")
)

display(model_analysis)
```

```python
#Machine Count by Age Group
age_distribution = df.groupBy(
    "age_group"
).count()

display(age_distribution)
```
