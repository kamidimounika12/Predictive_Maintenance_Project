```python
#Load Failure Data
from pyspark.sql.functions import *

failures = spark.table(
    "workspace.predictive_maintenance.silver_failures"
)

display(failures.limit(10))
```

```python
#Failure Count by Component
failure_types = failures.groupBy(
    "failure"
).count().orderBy(
    col("count").desc()
)

display(failure_types)
```

```python
#Failure Percentage
total_failures = failures.count()

failure_percentage = failure_types.withColumn(
    "failure_percentage",
    round(
        (col("count") / total_failures) * 100,
        2
    )
)

display(failure_percentage)
```

```python
#Failure Trend Over Time
failure_trend = failures.withColumn(
    "year_month",
    date_format("datetime", "yyyy-MM")
).groupBy(
    "year_month",
    "failure"
).count()

display(failure_trend)
```
