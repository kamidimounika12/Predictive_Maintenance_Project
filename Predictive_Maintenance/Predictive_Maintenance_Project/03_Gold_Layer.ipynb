```python
#Load Silver Tables
telemetry = spark.table("workspace.predictive_maintenance.silver_telemetry")

machines = spark.table("workspace.predictive_maintenance.silver_machines")
```

```python
#Gold Table 1: Machine Telemetry
gold_machine_telemetry = telemetry.join(
    machines,
    "machineID",
    "left"
)
```

```python
#Check
display(gold_machine_telemetry.limit(10))
```

```python
#Verify Count
print(
    "Gold Machine Telemetry Count:",
    gold_machine_telemetry.count()
)
```

```python
#Save Gold Table
gold_machine_telemetry.write \
.format("delta") \
.mode("overwrite") \
.saveAsTable(
    "workspace.predictive_maintenance.gold_machine_telemetry"
)
```

```python
#Verify Table
spark.sql("""
SHOW TABLES IN workspace.predictive_maintenance
""").show(truncate=False)
```

```python
#Verify Saved Data
display(
    spark.table(
        "workspace.predictive_maintenance.gold_machine_telemetry"
    ).limit(10)
)
```
