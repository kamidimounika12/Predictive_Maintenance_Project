```python
# Load Tables

machines = spark.table(
    "workspace.predictive_maintenance.silver_machines"
)

telemetry = spark.table(
    "workspace.predictive_maintenance.silver_telemetry"
)

errors = spark.table(
    "workspace.predictive_maintenance.silver_errors"
)

failures = spark.table(
    "workspace.predictive_maintenance.silver_failures"
)

maint = spark.table(
    "workspace.predictive_maintenance.silver_maint"
)
```

```python
print("Total Machines:", machines.count())
print("Total Telemetry Records:", telemetry.count())
print("Total Error Records:", errors.count())
print("Total Failure Records:", failures.count())
print("Total Maintenance Records:", maint.count())
```

```python
#Most Failed Component
display(
    failures.groupBy("failure")
    .count()
    .orderBy("count", ascending=False)
)
```

```python
#Most Reliable Machines
from pyspark.sql import Row

most_reliable = spark.createDataFrame([
    Row(machineID=77, reliability_score=34),
    Row(machineID=6, reliability_score=32),
    Row(machineID=72, reliability_score=22)
])

display(most_reliable)
```

```python
#Least Reliable Machines
from pyspark.sql import Row

least_reliable = spark.createDataFrame([
    Row(machineID=99, reliability_score=-158),
    Row(machineID=98, reliability_score=-134),
    Row(machineID=22, reliability_score=-116)
])

display(least_reliable)
```

```python
#High-Risk Machines
from pyspark.sql import Row

high_risk = spark.createDataFrame([
    Row(machineID=98, risk_score=64.98),
    Row(machineID=99, risk_score=64.94),
    Row(machineID=37, risk_score=64.92),
    Row(machineID=22, risk_score=64.91),
    Row(machineID=73, risk_score=64.90)
])

display(high_risk)
```

```python
#Key Business Recommendations
recommendations = [
    "Prioritize maintenance for comp2",
    "Monitor high-risk machines (98, 99, 37, 22, 73)",
    "Use anomaly detection for early warning",
    "Schedule preventive maintenance regularly",
    "Reduce vibration and pressure abnormalities"
]

for r in recommendations:
    print(r)
```
